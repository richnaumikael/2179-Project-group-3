```python
# ============================================================
# TRANSIT DELAY VS WEATHER (2024)
# UNIFIED ANALYSIS
#
# Includes:
#   1. Load and merge modelling + SMHI data
#   2. Calculate 24h / daily change variables
#   3. Fixed correlation matrix for ALL lines
#   4. Scatter/dot plots of every variable vs median_delay
#   5. Categorical boxplots with decoded Rådande väder labels
#   6. 64 -> 32 neural network
#   7. XGBoost regression
#   8. K-means clustering
#   9. Three k-values with highest silhouette scores
#  10. Cluster composition:
#       - snow depth
#       - precipitation
#       - temperature
#       - freezing
#       - humidity
#       - hour
#       - weekday
#       - month
#       - individual line
#
# IMPORTANT:
# The zero values in precipitation, precip_3h, frozen_precip
# and snow_depth_cm are removed ONLY from the correlation/
# plotting analysis.
#
# The zeros remain in the ML models because zero precipitation
# or zero snow is a real observation and should not normally
# be treated as missing in a predictive model.
# ============================================================


# %% 1. IMPORTS AND SETTINGS

import os
import textwrap
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")


MODELLING_FILE = "modelling_data_2024.csv"

SOIL_FILE = (
    "(markens tillstånd_dag Vallentuna)"
    " smhi-opendata_40_98310_202401_202412.csv"
)

WEATHER_FILE = (
    "(rådande väder Bromma)"
    " smhi-opendata_13_97200_202401_202412.csv"
)

HUMIDITY_FILE = (
    "(relativ luftfuktighet Stockholm-observatoriekullen)"
    " smhi-opendata_6_98230_202401_202412.csv"
)

TARGET = "median_delay"

LINES = [
    "BUS 1",
    "BUS 4",
    "METRO 10",
    "METRO 19",
    "TRAIN 43",
    "TRAM 30"
]

# Remove the closest observation to 1800 seconds/minutes
# from the categorical boxplots only.
BOXPLOT_OUTLIER_TARGET = 1800
BOXPLOT_OUTLIER_TOLERANCE = 100

# Clustering
K_RANGE = range(2, 13)
TOP_N_CLUSTER_COUNTS = 3
RANDOM_STATE = 0


# %% 2. HELPERS

def read_smhi(path):
    """
    Read an SMHI CSV and automatically locate the row
    beginning with 'Datum'.
    """

    with open(path, encoding="utf-8-sig") as f:
        header_row = next(
            i for i, line in enumerate(f)
            if line.startswith("Datum")
        )

    out = pd.read_csv(
        path,
        skiprows=header_row,
        sep=None,
        engine="python",
        encoding="utf-8-sig"
    )

    out.columns = out.columns.str.strip()

    return out


def utc_to_local(date_col, time_col):
    """
    Convert SMHI UTC timestamp to Europe/Stockholm.
    """

    utc = pd.to_datetime(
        date_col.astype(str) + " " + time_col.astype(str),
        errors="coerce"
    )

    return (
        utc.dt.tz_localize("UTC")
        .dt.tz_convert("Europe/Stockholm")
        .dt.tz_localize(None)
    )


def delta_ignore_zero(series):
    """
    Difference from the previous NON-ZERO value.

    Used for snow/precipitation type variables where zero
    means no observed accumulation.
    """

    nonzero = (
        series
        .replace(0, np.nan)
        .dropna()
    )

    return nonzero.diff().reindex(series.index)


def numeric_code(value):
    try:
        return int(float(value))
    except (ValueError, TypeError):
        return None


# %% 3. RÅDANDE VÄDER CODE MAP
#
# This is the mapping supplied in the parameter guide.
# Labels are deliberately descriptions rather than numerical
# codes on the boxplot x-axis.

RADANDE_VADER_MAP = {

    0: "Molnens utveckling har icke kunnat observeras eller icke observerats",
    1: "Moln har upplösts helt eller avtagit i utsträckning, i mäktighet eller i täthet",
    2: "Molnhimlen i stort sett oförändrad",
    3: "Moln har bildats eller tilltagit i utsträckning, i mäktighet eller i täthet",
    4: "Sikten nedsatt av brandrök eller fabriksrök",
    5: "Torrdis (solrök)",
    6: "Stoft svävar i luften, men uppvirvlas ej av vinden vid observationsterminen",
    7: "Stoft eller sand uppvirvlas av vinden men ingen utpräglad sandvirvel och ingen sandstorm inom synhåll",
    8: "Utpräglad stoft- eller sandvirvel vid observationsterminen eller under senaste timmen men ingen sandstorm",
    9: "Sandstorm under senaste timmen eller inom synhåll vid observationsterminen",
    10: "Fuktdis med sikt 1-10 km",
    11: "Låg dimma i bankar på stationen",
    12: "Mer eller mindre sammanhängande låg dimma på stationen",
    13: "Kornblixt",
    14: "Nederbörd inom synhåll, som ej når marken eller havsytan",
    15: "Nederbörd som når marken eller havsytan på avstånd större än 5 km",
    16: "Nederbörd som når marken eller havsytan på avstånd mindre än 5 km",
    17: "Åska vid observationsterminen men ingen nederbörd på stationen",
    18: "Utpräglade starka vindbyar",
    19: "Skydrag eller tromb",
    20: "Duggregn eller kornsnö under senaste timmen men ej vid observationsterminen",
    21: "Regn under senaste timmen men ej vid observationsterminen",
    22: "Snöfall under senaste timmen men ej vid observationsterminen",
    23: "Snöblandat regn eller iskorn under senaste timmen men ej vid observationsterminen",
    24: "Underkylt regn eller duggregn under senaste timmen men ej vid observationsterminen",
    25: "Regnskurar under senaste timmen men ej vid observationsterminen",
    26: "Byar av snö eller snöblandat regn under senaste timmen men ej vid observationsterminen",
    27: "Byar av hagel med eller utan regn under senaste timmen men ej vid observationsterminen",
    28: "Dimma under senaste timmen men ej vid observationsterminen",
    29: "Åska med eller utan nederbörd under senaste timmen men ej vid observationsterminen",
    30: "Lätt eller måttlig sandstorm har avtagit i styrka",
    31: "Lätt eller måttlig sandstorm utan märkbar förändring",
    32: "Lätt eller måttlig sandstorm har börjat eller tilltagit i styrka",
    33: "Kraftig sandstorm har avtagit i styrka",
    34: "Kraftig sandstorm utan märkbar förändring",
    35: "Kraftig sandstorm har börjat eller tilltagit i styrka",
    36: "Lågt och lätt eller måttligt snödrev",
    37: "Lågt men tätt snödrev",
    38: "Högt men lätt eller måttligt snödrev",
    39: "Högt och tätt snödrev",
    40: "Dimma inom synhåll vid observationsterminen, nående över ögonhöjd",
    41: "Dimma i bankar på stationen",
    42: "Dimma, med skymt av himlen, har blivit lättare",
    43: "Dimma, utan skymt av himlen, har blivit lättare",
    44: "Dimma, med skymt av himlen, oförändrad",
    45: "Dimma, utan skymt av himlen, oförändrad",
    46: "Dimma, med skymt av himlen, har börjat eller tätnat",
    47: "Dimma, utan skymt av himlen, har börjat eller tätnat",
    48: "Underkyld dimma, med skymt av himlen",
    49: "Underkyld dimma, utan skymt av himlen",

    50: "Lätt duggregn med avbrott",
    51: "Lätt duggregn, ihållande",
    52: "Måttligt duggregn med avbrott",
    53: "Måttligt duggregn, ihållande",
    54: "Tätt duggregn med avbrott",
    55: "Tätt duggregn, ihållande",
    56: "Lätt underkylt duggregn",
    57: "Måttligt eller tätt underkylt duggregn",
    58: "Lätt duggregn tillsammans med regn",
    59: "Måttligt eller tätt duggregn tillsammans med regn",

    60: "Lätt regn med avbrott",
    61: "Lätt regn, ihållande",
    62: "Måttligt regn med avbrott",
    63: "Måttligt regn, ihållande",
    64: "Starkt regn med avbrott",
    65: "Starkt regn ihållande",
    66: "Lätt underkylt regn",
    67: "Måttligt eller starkt underkylt regn",
    68: "Lätt regn eller duggregn tillsammans med snö",
    69: "Måttligt eller starkt regn eller duggregn tillsammans med snö",

    70: "Lätt snöfall med avbrott",
    71: "Lätt snöfall, ihållande",
    72: "Måttligt snöfall med avbrott",
    73: "Måttligt snöfall, ihållande",
    74: "Tätt snöfall med avbrott",
    75: "Tätt snöfall, ihållande",
    76: "Isnålar",
    77: "Kornsnö",
    78: "Enstaka snöstjärnor",
    79: "Iskorn",

    80: "Lätta regnskurar",
    81: "Måttliga eller kraftiga regnskurar",
    82: "Mycket kraftiga regnskurar (skyfall)",
    83: "Lätt snöblandat regn i byar",
    84: "Måttligt eller kraftigt snöblandat regn i byar",
    85: "Lätta snöbyar",
    86: "Måttliga eller kraftiga snöbyar",
    87: "Lätta byar av småhagel eller snöhagel",
    88: "Måttliga eller kraftiga byar av småhagel eller snöhagel",
    89: "Lätta byar av ishagel utan åska",
    90: "Måttliga eller kraftiga byar av ishagel utan åska",

    91: "Lätt regn vid observationsterminen, åskväder under senaste timmen",
    92: "Måttligt eller starkt regn vid observationsterminen, åskväder under senaste timmen",
    93: "Lätt snöfall, snöblandat regn eller hagel vid observationsterminen, åskväder senaste timmen",
    94: "Måttligt eller starkt snöfall, snöblandat regn eller hagel, åskväder senaste timmen",

    95: "Svagt eller måttligt åskväder utan hagel men med regn eller snö",
    96: "Svagt eller måttligt åskväder med hagel",
    97: "Kraftigt åskväder utan hagel men med regn eller snö",
    98: "Kraftigt åskväder med sandstorm",
    99: "Kraftigt åskväder med hagel",

    100: "Inget signifikant väder observerat",
    101: "Moln har upplösts helt eller avtagit under senaste timmen",
    102: "Molnhimlen i stort sett oförändrad under senaste timmen",
    103: "Moln har bildats eller tilltagit under senaste timmen",
    104: "Dis eller rök eller stoft i luften, sikt >= 1 km",
    105: "Dis eller rök eller stoft i luften, sikt < 1 km",

    110: "Fuktdis med sikt 1-10 km",
    111: "Isnålar",
    112: "Blixt på avstånd",
    118: "Utpräglade starka vindbyar",

    120: "Dimma",
    121: "Nederbörd",
    122: "Duggregn eller kornsnö",
    123: "Regn",
    124: "Snöfall",
    125: "Underkylt duggregn eller regn",
    126: "Åskväder med eller utan nederbörd",
    127: "Snödrev eller sandstorm",
    128: "Snödrev eller sandstorm, sikt >= 1 km",
    129: "Snödrev eller sandstorm, sikt < 1 km",
    130: "Dimma",
    131: "Dimma i bankar på stationen",
    132: "Dimma har blivit lättare under senaste timmen",
    133: "Dimma oförändrad under senaste timmen",
    134: "Dimma har börjat eller tätnat under senaste timmen",
    135: "Underkyld dimma",

    140: "Nederbörd",
    141: "Lätt eller måttlig nederbörd",
    142: "Kraftig nederbörd",
    143: "Flytande nederbörd, lätt eller måttlig",
    144: "Flytande nederbörd, kraftig",
    145: "Fast nederbörd, lätt eller måttlig",
    146: "Fast nederbörd, kraftig",
    147: "Lätt eller måttlig underkyld nederbörd",
    148: "Kraftig underkyld nederbörd",

    150: "Duggregn",
    151: "Lätt duggregn",
    152: "Måttligt duggregn",
    153: "Tätt duggregn",
    154: "Lätt underkylt duggregn",
    155: "Måttligt underkylt duggregn",
    156: "Tätt underkylt duggregn",
    157: "Lätt duggregn tillsammans med regn",
    158: "Måttligt eller tätt duggregn tillsammans med regn",

    160: "Regn",
    161: "Lätt regn",
    162: "Måttligt regn",
    163: "Starkt regn",
    164: "Lätt underkylt regn",
    165: "Måttligt underkylt regn",
    166: "Starkt underkylt regn",
    167: "Lätt regn eller duggregn tillsammans med snö",
    168: "Måttligt eller starkt regn eller duggregn tillsammans med snö",

    170: "Snöfall",
    171: "Lätt snöfall",
    172: "Måttligt snöfall",
    173: "Tätt snöfall",
    174: "Lätt småhagel",
    175: "Måttligt småhagel",
    176: "Kraftigt småhagel",
    177: "Kornsnö",
    178: "Isnålar",

    180: "Regnskurar",
    181: "Lätta regnskurar",
    182: "Måttliga regnskurar",
    183: "Kraftiga regnskurar",
    184: "Mycket kraftiga regnskurar (skyfall)",
    185: "Lätta snöbyar",
    186: "Måttliga snöbyar",
    187: "Kraftiga snöbyar",
    189: "Hagel",

    190: "Åskväder",
    191: "Svagt eller måttligt åskväder utan nederbörd",
    192: "Svagt eller måttligt åskväder med regnskurar eller snöbyar",
    193: "Svagt eller måttligt åskväder med hagel",
    194: "Kraftigt åskväder utan nederbörd",
    195: "Kraftigt åskväder med regnskurar eller snöbyar",
    196: "Kraftigt åskväder med hagel",
    199: "Tromb eller tornado",

    204: "Vulkanaska som spridits högt upp i luften",
    206: "Tjockt stoftdis, sikt < 1 km",
    207: "Vattenstänk vid station pga blåst",
    208: "Drivande stoft eller sand",
    209: "Kraftig stoft- eller sandstorm på avstånd",
    210: "Snödis",
    211: "Snöstorm eller kraftigt snödrev som ger extremt dålig sikt",
    213: "Blixt mellan moln och marken",
    217: "Åska utan regnskur",
    219: "Tromb eller tornado vid stationen eller inom synhåll under senaste timmen",

    220: "Avlagring av vulkanaska",
    221: "Avlagring av stoft eller sand",
    222: "Dagg",
    223: "Utfällning av blöt snö",
    224: "Lätt eller måttlig dimfrost",
    225: "Kraftig dimfrost",
    226: "Rimfrost",
    227: "Kraftig isbeläggning pga underkyld nederbörd",
    228: "Isskorpa",

    230: "Stoft- eller sandstorm med temperatur under fryspunkten",
    239: "Kraftigt snödrev och/eller snöfall",

    241: "Dimma till havs",
    242: "Dimma i dalgång",
    243: "Sjörök i Arktis eller vid Antarktis",
    244: "Advektionsdimma över vatten",
    245: "Advektionsdimma över land",
    246: "Dimma över is eller snö",
    247: "Tät dimma, sikt 60-90 m",
    248: "Tät dimma, sikt 30-60 m",
    249: "Tät dimma, sikt < 30 m",

    250: "Duggregn, intensitet < 0,10 mm/timme",
    251: "Duggregn, intensitet 0,10-0,19 mm/timme",
    252: "Duggregn, intensitet 0,20-0,39 mm/timme",
    253: "Duggregn, intensitet 0,40-0,79 mm/timme",
    254: "Duggregn, intensitet 0,80-1,59 mm/timme",
    255: "Duggregn, intensitet 1,60-3,19 mm/timme",
    256: "Duggregn, intensitet 3,20-6,39 mm/timme",
    257: "Duggregn, intensitet > 6,40 mm/timme",

    259: "Duggregn och snöfall",

    260: "Regn, intensitet < 1,0 mm/timme",
    261: "Regn, intensitet 1,0-1,9 mm/timme",
    262: "Regn, intensitet 2,0-3,9 mm/timme",
    263: "Regn, intensitet 4,0-7,9 mm/timme",
    264: "Regn, intensitet 8,0-15,9 mm/timme",
    265: "Regn, intensitet 16,0-31,9 mm/timme",
    266: "Regn, intensitet 32,0-63,9 mm/timme",
    267: "Regn, intensitet > 64,0 mm/timme",

    270: "Snö, intensitet < 1,0 cm/timme",
    271: "Snö, intensitet 1,0-1,9 cm/timme",
    272: "Snö, intensitet 2,0-3,9 cm/timme",
    273: "Snö, intensitet 4,0-7,9 cm/timme",
    274: "Snö, intensitet 8,0-15,9 cm/timme",
    275: "Snö, intensitet 16,0-31,9 cm/timme",
    276: "Snö, intensitet 32,0-63,9 cm/timme",
    277: "Snö, intensitet > 64,0 cm/timme",

    278: "Snöfall eller isnålar från klar himmel",
    279: "Frysande blötsnö",
    280: "Regn",
    281: "Underkylt regn",
    282: "Snöblandat regn",
    283: "Snöfall",
    284: "Småhagel eller snöhagel",
    285: "Småhagel eller snöhagel tillsammans med regn",
    286: "Småhagel eller snöhagel tillsammans med snöblandat regn",
    287: "Småhagel eller snöhagel tillsammans med snö",
    288: "Hagel",
    289: "Hagel tillsammans med regn",
    290: "Hagel tillsammans med snöblandat regn",
    291: "Hagel tillsammans med snö",
    292: "Skurar eller åska till havs",
    293: "Skurar eller åska över berg",

    508: "Inga signifikanta fenomen att rapportera",
    509: "Ingen observation, data ej tillgängligt",
    510: "Rådande och gammalt väder saknas men förväntades",
    511: "Saknat värde"
}


# %% 4. MARKENS TILLSTÅND MAP
#
# The supplied parameter table contains Rådande väder, but does
# not contain a corresponding Markens tillstånd code table.
#
# If your soil-condition codes are already text, they are used
# directly.
#
# If they are numerical, insert the corresponding mapping here.
#
# Example:
#
# MARKENS_TILLSTAND_MAP = {
#     0: "....",
#     1: "....",
# }
#
MARKENS_TILLSTAND_MAP = {}


# %% 5. LOAD MODELLING DATA

df = pd.read_csv(MODELLING_FILE)
df.columns = df.columns.str.strip()

df["date_dt"] = pd.to_datetime(
    df["date"].astype(str),
    format="%Y%m%d"
)

df["local_timestamp"] = (
    df["date_dt"]
    + pd.to_timedelta(df["hour"], unit="h")
)

df["local_date"] = df["date_dt"].dt.date


# %% 6. LOAD SMHI DATA

# Soil
soil = read_smhi(SOIL_FILE).rename(
    columns={
        "Datum": "local_date",
        "Markens tillstånd": "markens_tillstand"
    }
)

soil["local_date"] = (
    pd.to_datetime(soil["local_date"])
    .dt.date
)

soil["markens_tillstand"] = (
    soil["markens_tillstand"]
    .astype(str)
    .str.strip()
)

soil = (
    soil[
        [
            "local_date",
            "markens_tillstand"
        ]
    ]
    .drop_duplicates("local_date")
)


# Rådande väder
weather = read_smhi(WEATHER_FILE)

weather["local_timestamp"] = utc_to_local(
    weather["Datum"],
    weather["Tid (UTC)"]
)

weather["radande_vader"] = (
    weather["Rådande väder"]
    .astype(str)
    .str.strip()
)

weather = (
    weather
    .groupby(
        "local_timestamp",
        as_index=False
    )["radande_vader"]
    .agg(lambda x: x.mode().iloc[0])
)


# Humidity
humidity = read_smhi(HUMIDITY_FILE)

humidity["local_timestamp"] = utc_to_local(
    humidity["Datum"],
    humidity["Tid (UTC)"]
)

humidity["humidity"] = pd.to_numeric(
    humidity["Relativ Luftfuktighet"],
    errors="coerce"
)

humidity = (
    humidity
    .groupby(
        "local_timestamp",
        as_index=False
    )["humidity"]
    .mean()
)


# %% 7. MERGE

df = (
    df
    .merge(
        weather,
        on="local_timestamp",
        how="left",
        validate="many_to_one"
    )
    .merge(
        humidity,
        on="local_timestamp",
        how="left",
        validate="many_to_one"
    )
    .merge(
        soil,
        on="local_date",
        how="left",
        validate="many_to_one"
    )
)

print("Observations:", len(df))

print("\nMissing SMHI values:")
print(
    df[
        [
            "radande_vader",
            "humidity",
            "markens_tillstand"
        ]
    ].isna().sum()
)


# %% 8. CALCULATE DELTA VARIABLES

unique_hourly = (
    df[
        [
            "local_timestamp",
            "temperature",
            "humidity",
            "precip_24h",
            "frozen_precip"
        ]
    ]
    .drop_duplicates("local_timestamp")
    .sort_values("local_timestamp")
)


# d_precipitation_24h
precip_24h = (
    unique_hourly
    .set_index("local_timestamp")["precip_24h"]
)

df["d_precipitation_24h"] = (
    df["local_timestamp"]
    .map(delta_ignore_zero(precip_24h))
)


# d_frozen_precipitation_24h
frozen_precip = (
    unique_hourly
    .set_index("local_timestamp")["frozen_precip"]
)

df["d_frozen_precipitation_24h"] = (
    df["local_timestamp"]
    .map(
        frozen_precip.diff(24)
    )
)


# d_temperature_24h
temperature = (
    unique_hourly
    .set_index("local_timestamp")["temperature"]
)

df["d_temperature_24h"] = (
    df["local_timestamp"]
    .map(
        temperature.diff(24)
    )
)


# d_humidity_24h
humidity_series = (
    unique_hourly
    .set_index("local_timestamp")["humidity"]
)

df["d_humidity_24h"] = (
    df["local_timestamp"]
    .map(
        humidity_series.diff(24)
    )
)


# d_snow_depth
daily_snow = (
    df
    .groupby("local_date")["snow_depth_cm"]
    .mean()
    .sort_index()
)

df["d_snow_depth"] = (
    df["local_date"]
    .map(
        delta_ignore_zero(daily_snow)
    )
)


# %% 9. FIXED CORRELATION VARIABLES

CORR_VARS = [
    "median_delay",
    "mean_delay",
    "share_late",
    "month",
    "weekday",
    "hour",
    "snow_depth_cm",
    "temperature",
    "precipitation",
    "precip_3h",
    "wind",
    "frozen_precip",
    "is_freezing",
    "humidity",
    "d_precipitation_24h",
    "d_frozen_precipitation_24h",
    "d_snow_depth",
    "d_temperature_24h",
    "d_humidity_24h"
]


# Check
missing = [
    col
    for col in CORR_VARS
    if col not in df.columns
]

if missing:
    raise KeyError(
        "Missing correlation columns:\n"
        + "\n".join(missing)
    )


# Numeric conversion
for col in CORR_VARS:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )


# %% 10. ANALYSIS COPY — REMOVE ZERO VALUES

# IMPORTANT:
# Only the analysis/plotting copy is changed here.
# The original df retains the zeros for ML.

analysis_df = df.copy()

ZERO_TO_REMOVE = [
    "frozen_precip",
    "precip_3h",
    "precipitation",
    "snow_depth_cm"
]

for col in ZERO_TO_REMOVE:
    analysis_df[col] = (
        analysis_df[col]
        .replace(0, np.nan)
    )

print("\nZero values removed from correlation/plot analysis:")

for col in ZERO_TO_REMOVE:
    print(
        f"{col}: "
        f"{df[col].eq(0).sum()} zeros removed"
    )


# %% 11. FIND ~1800 OUTLIER FOR BOXPLOTS

valid_delay = analysis_df[
    TARGET
].dropna()

closest_idx = (
    (valid_delay - BOXPLOT_OUTLIER_TARGET)
    .abs()
    .idxmin()
)

closest_value = (
    analysis_df
    .loc[
        closest_idx,
        TARGET
    ]
)

if (
    abs(
        closest_value
        - BOXPLOT_OUTLIER_TARGET
    )
    <= BOXPLOT_OUTLIER_TOLERANCE
):

    BOX_OUTLIER_LABEL = {
        "index": closest_idx,
        "label": analysis_df.loc[
            closest_idx,
            "label"
        ],
        "timestamp": analysis_df.loc[
            closest_idx,
            "local_timestamp"
        ],
        "median_delay": closest_value
    }

    print(
        "\nRemoved from categorical boxplots:"
    )

    print(
        pd.Series(
            BOX_OUTLIER_LABEL
        )
    )

else:
    BOX_OUTLIER_LABEL = None

    print(
        "\nNo ~1800 outlier found within tolerance."
    )


# %% 12. DECODE CATEGORICAL VALUES

def decode_category(
    value,
    mapping,
    fallback=True
):

    code = numeric_code(value)

    if code is None:
        return str(value)

    if code in mapping:
        return mapping[code]

    if fallback:
        return f"Unknown code {code}"

    return str(value)


analysis_df["radande_vader_decoded"] = (
    analysis_df["radande_vader"]
    .map(
        lambda x:
        decode_category(
            x,
            RADANDE_VADER_MAP
        )
    )
)


analysis_df["markens_tillstand_decoded"] = (
    analysis_df["markens_tillstand"]
    .map(
        lambda x:
        decode_category(
            x,
            MARKENS_TILLSTAND_MAP
        )
    )
)


# %% 13. CORRELATION MATRICES + DOT PLOTS

def analyse_line(
    line_df,
    name
):

    numeric_df = (
        line_df[CORR_VARS]
        .apply(
            pd.to_numeric,
            errors="coerce"
        )
    )


    corr = numeric_df.corr(
        method="pearson"
    )


    # --------------------------------------------------------
    # Correlation matrix
    # --------------------------------------------------------

    plt.figure(
        figsize=(18, 15)
    )

    sns.heatmap(
        corr,
        annot=True,
        fmt=".2f",
        cmap="coolwarm",
        center=0,
        vmin=-1,
        vmax=1,
        square=True
    )

    plt.title(
        f"Pearson correlation matrix - {name}"
    )

    plt.tight_layout()
    plt.show()


    # --------------------------------------------------------
    # Correlation with median delay
    # --------------------------------------------------------

    print(
        f"\nCorrelation with median_delay - {name}:"
    )

    print(
        corr[
            TARGET
        ]
        .drop(TARGET)
        .sort_values(
            key=abs,
            ascending=False
        )
        .round(3)
        .to_string()
    )


    # --------------------------------------------------------
    # DOT / SCATTER MAP
    # --------------------------------------------------------

    plot_vars = [
        col
        for col in CORR_VARS
        if col != TARGET
    ]

    ncols = 4

    nrows = int(
        np.ceil(
            len(plot_vars)
            / ncols
        )
    )

    fig, axes = plt.subplots(
        nrows,
        ncols,
        figsize=(
            5 * ncols,
            4 * nrows
        )
    )

    axes = np.atleast_1d(
        axes
    )


    for ax, var in zip(
        axes.flat,
        plot_vars
    ):

        plot_data = (
            line_df[
                [
                    var,
                    TARGET
                ]
            ]
            .dropna()
        )


        sns.scatterplot(
            data=plot_data,
            x=var,
            y=TARGET,
            alpha=0.40,
            s=15,
            ax=ax
        )


        # Keep the regression line from the previous version.
        if (
            len(plot_data) > 2
            and plot_data[var].nunique() > 1
        ):

            sns.regplot(
                data=plot_data,
                x=var,
                y=TARGET,
                scatter=False,
                ax=ax
            )


        ax.set_title(
            f"{var} vs median_delay"
        )


    for ax in axes.flat[
        len(plot_vars):
    ]:
        ax.axis("off")


    fig.suptitle(
        f"Variables vs median_delay - {name}",
        y=1.0,
        fontsize=15
    )

    plt.tight_layout()
    plt.show()


# All lines
analyse_line(
    analysis_df,
    "ALL LINES"
)


# Individual lines
for line in LINES:

    analyse_line(
        analysis_df[
            analysis_df["label"] == line
        ].copy(),
        line
    )


# %% 14. CATEGORICAL BOXPLOTS

def categorical_boxplots(
    line_df,
    name
):

    plot_df = line_df.copy()


    # --------------------------------------------------------
    # Remove the ~1800 observation
    # --------------------------------------------------------

    if BOX_OUTLIER_LABEL is not None:

        plot_df = plot_df[
            ~(
                (
                    plot_df["label"]
                    == BOX_OUTLIER_LABEL["label"]
                )
                &
                (
                    plot_df["local_timestamp"]
                    == BOX_OUTLIER_LABEL["timestamp"]
                )
            )
        ].copy()


    # --------------------------------------------------------
    # Wrapped decoded labels
    # --------------------------------------------------------

    plot_df["weather_label"] = (
        plot_df["radande_vader"]
        .map(
            lambda x:
            decode_category(
                x,
                RADANDE_VADER_MAP
            )
        )
    )


    plot_df["soil_label"] = (
        plot_df["markens_tillstand"]
        .map(
            lambda x:
            decode_category(
                x,
                MARKENS_TILLSTAND_MAP
            )
        )
    )


    # --------------------------------------------------------
    # Category orders
    # --------------------------------------------------------

    weather_codes = (
        plot_df["radande_vader"]
        .dropna()
        .map(numeric_code)
        .dropna()
        .astype(int)
        .unique()
    )

    weather_codes = sorted(
        set(weather_codes)
    )


    weather_order = [
        RADANDE_VADER_MAP.get(
            code,
            f"Unknown code {code}"
        )
        for code in weather_codes
    ]


    soil_values = (
        plot_df["soil_label"]
        .dropna()
        .unique()
        .tolist()
    )


    # --------------------------------------------------------
    # Make wide enough for long labels
    # --------------------------------------------------------

    fig, axes = plt.subplots(
        1,
        2,
        figsize=(24, 8)
    )


    # --------------------------------------------------------
    # Markens tillstånd
    # --------------------------------------------------------

    soil_plot = (
        plot_df[
            [
                "soil_label",
                TARGET
            ]
        ]
        .dropna()
    )


    if len(soil_plot) > 0:

        sns.boxplot(
            data=soil_plot,
            x="soil_label",
            y=TARGET,
            order=soil_values,
            ax=axes[0]
        )

        axes[0].set_title(
            f"Markens tillstånd vs median_delay - {name}"
        )

        axes[0].set_xlabel(
            "Markens tillstånd"
        )

        axes[0].tick_params(
            axis="x",
            rotation=75
        )

    else:

        axes[0].set_axis_off()


    # --------------------------------------------------------
    # Rådande väder
    # --------------------------------------------------------

    weather_plot = (
        plot_df[
            [
                "weather_label",
                TARGET
            ]
        ]
        .dropna()
    )


    if len(weather_plot) > 0:

        sns.boxplot(
            data=weather_plot,
            x="weather_label",
            y=TARGET,
            order=weather_order,
            ax=axes[1]
        )

        axes[1].set_title(
            f"Rådande väder vs median_delay - {name}"
        )

        axes[1].set_xlabel(
            "Rådande väder"
        )

        axes[1].tick_params(
            axis="x",
            rotation=90
        )

    else:

        axes[1].set_axis_off()


    plt.tight_layout()
    plt.show()


# All lines
categorical_boxplots(
    analysis_df,
    "ALL LINES"
)


# Individual lines
for line in LINES:

    categorical_boxplots(
        analysis_df[
            analysis_df["label"] == line
        ].copy(),
        line
    )


# %% 14B. AVERAGE / MEDIAN DELAY PER LINE

delay_summary = (
    df
    .groupby("label")
    .agg(
        average_delay=("mean_delay", "mean"),
        median_delay=("median_delay", "median"),
        n_observations=("median_delay", "count")
    )
    .reindex(LINES)
    .round(2)
)

print("\nAVERAGE / MEDIAN DELAY PER LINE")
print(delay_summary.to_string())


# %% 14C. ARE THE LINES' DELAYS CORRELATED WITH EACH OTHER?
#
# Pivot so each row is one point in time (local_timestamp) and each
# column is one line's median_delay at that time, then correlate the
# columns. This checks whether lines tend to be delayed together
# (e.g. a shared disruption) rather than independently.

delay_wide = df.pivot_table(
    index="local_timestamp",
    columns="label",
    values=TARGET
).reindex(columns=LINES)

line_delay_corr = delay_wide.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    line_delay_corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    square=True
)
plt.title("Correlation of median_delay BETWEEN lines")
plt.tight_layout()
plt.show()

print("\nCorrelation of median_delay between lines:")
print(line_delay_corr.round(3).to_string())


# %% 15. SAVE MERGED DATA

df.to_csv(
    "modelling_data_2024_with_SMHI.csv",
    index=False
)

print(
    "\nSaved:"
    " modelling_data_2024_with_SMHI.csv"
)



# %% 16. MACHINE LEARNING: SHARED SETUP
#
# Models:
#   1. Linear Regression
#   2. Neural Network (64 -> 32 -> 1)
#   3. XGBoost
#
# Evaluation:
#   - Chronological 70/15/15 train/validation/test split
#   - Same test observations and preprocessing for all models
#   - One pooled model using all lines
#   - One independent model per line
#   - Final R² comparison table
#
# "Overall pooled" means one model trained on all lines.
# "Overall separate models" means combining test predictions
# from the six independently trained line-specific models.
# ============================================================

from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)

from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.layers import Dense, Dropout, Input
from tensorflow.keras.models import Sequential

# Install XGBoost if necessary
try:
    from xgboost import XGBRegressor
except ImportError:
    print("Installing XGBoost...")
    import sys
    import subprocess

    subprocess.check_call([
        sys.executable, "-m", "pip", "install", "-q", "xgboost"
    ])

    from xgboost import XGBRegressor


# Numeric predictors shared by all models.
# No delay measurements from other lines are used as predictors.
model_num = [
    "n_stops",
    "temp_tor",
    "regn_tor",
    "wind_tor",
    "snow_depth_cm",
    "temperature",
    "precipitation",
    "wind",
    "precip_3h",
    "precip_24h",
    "frozen_precip",
    "is_freezing",
    "humidity",
    "d_precipitation_24h",
    "d_frozen_precipitation_24h",
    "d_snow_depth",
    "d_temperature_24h",
    "d_humidity_24h",
    "hour",
    "weekday",
    "is_weekend",
    "month"
]

# The pooled model can use the line identity as a categorical feature.
# A separate line model does not need this feature.
model_cat_pooled = [
    "markens_tillstand",
    "radande_vader",
    "label"
]

model_cat_line = [
    "markens_tillstand",
    "radande_vader"
]


# %% 17. PLOTTING HELPER

def actual_vs_predicted_plot(actual, predicted, title, r2):

    actual = np.asarray(actual)
    predicted = np.asarray(predicted)

    plt.figure(figsize=(7, 7))
    sns.scatterplot(
        x=actual,
        y=predicted,
        alpha=0.45,
        s=25
    )

    min_value = min(actual.min(), predicted.min())
    max_value = max(actual.max(), predicted.max())

    plt.plot(
        [min_value, max_value],
        [min_value, max_value],
        linestyle="--"
    )

    plt.xlabel("Actual median delay")
    plt.ylabel("Predicted median delay")
    plt.title(title)

    plt.text(
        0.05,
        0.95,
        f"$R^2 = {r2:.3f}$",
        transform=plt.gca().transAxes,
        verticalalignment="top",
        fontsize=14
    )

    plt.tight_layout()
    plt.show()


# %% 18. TRAIN AND EVALUATE ALL THREE MODELS

def train_evaluate(data_subset, model_cat, tag):
    """
    Fit Linear Regression, Neural Network and XGBoost.

    All three models use:
      - The same chronological train/validation/test split
      - The same preprocessed input features
      - The same held-out test observations

    Returns metrics and test predictions so the six independent
    line models can also be evaluated together across the network.
    """

    data_subset = data_subset.copy()

    if "timestamp" not in data_subset.columns:
        data_subset["timestamp"] = data_subset["local_timestamp"]

    data_subset["timestamp"] = pd.to_datetime(
        data_subset["timestamp"],
        errors="coerce"
    )

    data_subset = (
        data_subset
        .sort_values("timestamp")
        .reset_index(drop=True)
    )

    # Convert numeric inputs and target safely.
    for col in model_num + [TARGET]:
        data_subset[col] = pd.to_numeric(
            data_subset[col],
            errors="coerce"
        )

        data_subset[col] = data_subset[col].replace(
            [np.inf, -np.inf],
            np.nan
        )

    # Rows without the target cannot be used for supervised learning.
    data_subset = (
        data_subset
        .dropna(subset=[TARGET])
        .reset_index(drop=True)
    )

    n = len(data_subset)

    i_train = int(n * 0.70)
    i_val = int(n * 0.85)

    train = data_subset.iloc[:i_train].copy()
    val = data_subset.iloc[i_train:i_val].copy()
    test = data_subset.iloc[i_val:].copy()

    print(
        f"\n[{tag}] train / validation / test: "
        f"{len(train)} / {len(val)} / {len(test)}"
    )

    if len(train) < 30 or len(val) < 2 or len(test) < 5:
        print(
            f"[{tag}] Not enough observations to train/evaluate "
            "reliably. Skipping."
        )
        return None

    feature_cols = model_num + model_cat

    # Preprocessing is fitted on training data only.
    preprocess = ColumnTransformer(
        transformers=[
            (
                "num",
                Pipeline([
                    ("impute", SimpleImputer(strategy="median")),
                    ("scale", StandardScaler())
                ]),
                model_num
            ),
            (
                "cat",
                Pipeline([
                    (
                        "impute",
                        SimpleImputer(
                            strategy="constant",
                            fill_value="missing"
                        )
                    ),
                    (
                        "onehot",
                        OneHotEncoder(
                            handle_unknown="ignore",
                            sparse_output=False
                        )
                    )
                ]),
                model_cat
            )
        ]
    )

    X_train = preprocess.fit_transform(train[feature_cols])
    X_val = preprocess.transform(val[feature_cols])
    X_test = preprocess.transform(test[feature_cols])

    y_train_original = train[TARGET].to_numpy()
    y_val_original = val[TARGET].to_numpy()
    y_true = test[TARGET].to_numpy()

    # --------------------------------------------------------
    # MODEL 1: LINEAR REGRESSION
    # --------------------------------------------------------
    #
    # Linear Regression does not require target scaling.
    # Input features use the same preprocessing as the other models.

    lr_model = LinearRegression()

    lr_model.fit(
        X_train,
        y_train_original
    )

    lr_pred = lr_model.predict(X_test)

    lr_mae = mean_absolute_error(y_true, lr_pred)
    lr_rmse = np.sqrt(mean_squared_error(y_true, lr_pred))
    lr_r2 = r2_score(y_true, lr_pred)

    print(
        f"[{tag}] Linear Regression - "
        f"MAE {lr_mae:.2f} | "
        f"RMSE {lr_rmse:.2f} | "
        f"R² {lr_r2:.3f}"
    )

    actual_vs_predicted_plot(
        y_true,
        lr_pred,
        f"{tag}: Linear Regression - actual vs predicted",
        lr_r2
    )

    # --------------------------------------------------------
    # MODEL 2: NEURAL NETWORK (64 -> 32 -> 1)
    # --------------------------------------------------------

    y_scaler = StandardScaler()

    y_train = y_scaler.fit_transform(
        train[[TARGET]]
    ).ravel()

    y_val = y_scaler.transform(
        val[[TARGET]]
    ).ravel()

    nn_model = Sequential([
        Input(shape=(X_train.shape[1],)),
        Dense(64, activation="relu"),
        Dropout(0.2),
        Dense(32, activation="relu"),
        Dense(1)
    ])

    nn_model.compile(
        optimizer="adam",
        loss="mse",
        metrics=["mae"]
    )

    nn_history = nn_model.fit(
        X_train,
        y_train,
        validation_data=(X_val, y_val),
        epochs=200,
        batch_size=32,
        verbose=0,
        callbacks=[
            EarlyStopping(
                monitor="val_loss",
                patience=15,
                restore_best_weights=True
            )
        ]
    )

    nn_pred_scaled = nn_model.predict(
        X_test,
        verbose=0
    ).reshape(-1, 1)

    nn_pred = y_scaler.inverse_transform(
        nn_pred_scaled
    ).ravel()

    nn_mae = mean_absolute_error(y_true, nn_pred)
    nn_rmse = np.sqrt(mean_squared_error(y_true, nn_pred))
    nn_r2 = r2_score(y_true, nn_pred)

    print(
        f"[{tag}] Neural Network    - "
        f"MAE {nn_mae:.2f} | "
        f"RMSE {nn_rmse:.2f} | "
        f"R² {nn_r2:.3f}"
    )

    actual_vs_predicted_plot(
        y_true,
        nn_pred,
        f"{tag}: Neural Network - actual vs predicted",
        nn_r2
    )

    plt.figure(figsize=(7, 4))
    plt.plot(
        nn_history.history["loss"],
        label="Training"
    )
    plt.plot(
        nn_history.history["val_loss"],
        label="Validation"
    )
    plt.xlabel("Epoch")
    plt.ylabel("MSE (scaled target)")
    plt.title(f"{tag}: neural network training history")
    plt.legend()
    plt.tight_layout()
    plt.show()

    # --------------------------------------------------------
    # MODEL 3: XGBOOST
    # --------------------------------------------------------

    xgb_model = XGBRegressor(
        objective="reg:squarederror",
        n_estimators=500,
        learning_rate=0.05,
        max_depth=4,
        min_child_weight=3,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_lambda=1.0,
        random_state=RANDOM_STATE,
        n_jobs=-1
    )

    xgb_model.fit(
        X_train,
        y_train_original
    )

    xgb_pred = xgb_model.predict(X_test)

    xgb_mae = mean_absolute_error(y_true, xgb_pred)
    xgb_rmse = np.sqrt(mean_squared_error(y_true, xgb_pred))
    xgb_r2 = r2_score(y_true, xgb_pred)

    print(
        f"[{tag}] XGBoost           - "
        f"MAE {xgb_mae:.2f} | "
        f"RMSE {xgb_rmse:.2f} | "
        f"R² {xgb_r2:.3f}"
    )

    actual_vs_predicted_plot(
        y_true,
        xgb_pred,
        f"{tag}: XGBoost - actual vs predicted",
        xgb_r2
    )

    # --------------------------------------------------------
    # XGBOOST FEATURE IMPORTANCE
    # --------------------------------------------------------

    feature_names = preprocess.get_feature_names_out()

    importance = pd.Series(
        xgb_model.feature_importances_,
        index=feature_names
    ).sort_values(ascending=False)

    print(f"\n[{tag}] Top XGBoost feature importances:")
    print(importance.head(15).round(4).to_string())

    plt.figure(figsize=(8, 6))
    importance.head(15).sort_values().plot(kind="barh")
    plt.xlabel("XGBoost importance")
    plt.title(f"{tag}: most important variables")
    plt.tight_layout()
    plt.show()

    # Return test predictions for both individual model results
    # and combined evaluation of all independent line models.
    return {
        "tag": tag,
        "n_test": len(test),

        "lr_mae": lr_mae,
        "lr_rmse": lr_rmse,
        "lr_r2": lr_r2,

        "nn_mae": nn_mae,
        "nn_rmse": nn_rmse,
        "nn_r2": nn_r2,

        "xgb_mae": xgb_mae,
        "xgb_rmse": xgb_rmse,
        "xgb_r2": xgb_r2,

        "y_true": y_true,
        "lr_pred": lr_pred,
        "nn_pred": nn_pred,
        "xgb_pred": xgb_pred,

        "top_features": importance.head(10)
    }


# %% 19. TRAIN POOLED MODEL AND SEPARATE MODEL FOR EACH LINE

model_results = {}

# ------------------------------------------------------------
# A. POOLED MODEL: ALL SIX LINES TOGETHER
# ------------------------------------------------------------
#
# One model of each algorithm is trained on the combined dataset.
# The line label is included as a categorical predictor.
#
# The delay of another line is NOT used as an input feature.

model_results["ALL LINES (pooled)"] = train_evaluate(
    df,
    model_cat_pooled,
    "ALL LINES (pooled)"
)


# ------------------------------------------------------------
# B. INDEPENDENT MODELS: ONE MODEL PER LINE
# ------------------------------------------------------------
#
# Each model is trained only on the observations for its own line.
# No other line's delay values are used.

for line in LINES:

    line_data = df[
        df["label"] == line
    ].copy()

    model_results[line] = train_evaluate(
        line_data,
        model_cat_line,
        line
    )


# %% 20. R² COMPARISON TABLE: POOLED AND INDIVIDUAL LINES

comparison_rows = []

for tag, res in model_results.items():

    if res is None:
        continue

    comparison_rows.append({
        "Dataset": tag,
        "Test observations": res["n_test"],
        "Linear Regression R²": res["lr_r2"],
        "Neural Network R²": res["nn_r2"],
        "XGBoost R²": res["xgb_r2"]
    })


r2_comparison = pd.DataFrame(comparison_rows)

print("\n" + "=" * 100)
print("R² COMPARISON: LINEAR REGRESSION VS NEURAL NETWORK VS XGBOOST")
print("=" * 100)

if not r2_comparison.empty:
    r2_comparison["Best model"] = (
        r2_comparison[
            [
                "Linear Regression R²",
                "Neural Network R²",
                "XGBoost R²"
            ]
        ]
        .idxmax(axis=1)
        .str.replace(" R²", "", regex=False)
    )

    display(
        r2_comparison.style.format({
            "Linear Regression R²": "{:.3f}",
            "Neural Network R²": "{:.3f}",
            "XGBoost R²": "{:.3f}"
        })
    )


# %% 21. OVERALL PERFORMANCE OF THE SIX SEPARATE LINE MODELS
#
# Concatenate the held-out test observations and predictions from
# the six independently trained models. Compute R² on the combined
# observations.
#
# This is different from the pooled model: each prediction here
# comes from a model trained only on that particular line.

separate_line_results = [
    model_results[line]
    for line in LINES
    if model_results.get(line) is not None
]

if separate_line_results:

    y_all_lines = np.concatenate([
        res["y_true"]
        for res in separate_line_results
    ])

    lr_all_lines = np.concatenate([
        res["lr_pred"]
        for res in separate_line_results
    ])

    nn_all_lines = np.concatenate([
        res["nn_pred"]
        for res in separate_line_results
    ])

    xgb_all_lines = np.concatenate([
        res["xgb_pred"]
        for res in separate_line_results
    ])

    separate_models_overall = {
        "Dataset": "ALL LINES (separate models combined)",
        "Test observations": len(y_all_lines),
        "Linear Regression R²": r2_score(
            y_all_lines, lr_all_lines
        ),
        "Neural Network R²": r2_score(
            y_all_lines, nn_all_lines
        ),
        "XGBoost R²": r2_score(
            y_all_lines, xgb_all_lines
        )
    }

    separate_models_overall["Best model"] = (
        pd.Series({
            "Linear Regression": separate_models_overall[
                "Linear Regression R²"
            ],
            "Neural Network": separate_models_overall[
                "Neural Network R²"
            ],
            "XGBoost": separate_models_overall[
                "XGBoost R²"
            ]
        })
        .idxmax()
    )

    # Add the combined independent-model result to the main table.
    r2_comparison = pd.concat(
        [
            r2_comparison,
            pd.DataFrame([separate_models_overall])
        ],
        ignore_index=True
    )

    print(
        "\nOVERALL R² FOR THE SIX INDEPENDENT LINE MODELS "
        "(combined test predictions):"
    )

    print(
        f"Linear Regression: "
        f"{separate_models_overall['Linear Regression R²']:.3f}"
    )

    print(
        f"Neural Network: "
        f"{separate_models_overall['Neural Network R²']:.3f}"
    )

    print(
        f"XGBoost: "
        f"{separate_models_overall['XGBoost R²']:.3f}"
    )


# Final table, including the separate-model combined result.
print("\nFINAL R² COMPARISON TABLE")

display(
    r2_comparison.style.format({
        "Linear Regression R²": "{:.3f}",
        "Neural Network R²": "{:.3f}",
        "XGBoost R²": "{:.3f}"
    })
)



# %% 20. CLUSTERING DATA

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import OneHotEncoder


cluster_src = (
    df[
        [
            "label",
            "local_date",
            "hour",
            "month",
            "weekday"
        ]
    ]
    .drop_duplicates(
        [
            "label",
            "local_date",
            "hour"
        ]
    )
    .copy()
)


# Cyclical hour
cluster_src["hour_sin"] = (
    np.sin(
        2 * np.pi
        * cluster_src["hour"]
        / 24
    )
)

cluster_src["hour_cos"] = (
    np.cos(
        2 * np.pi
        * cluster_src["hour"]
        / 24
    )
)


# Cyclical month
cluster_src["month_sin"] = (
    np.sin(
        2 * np.pi
        * cluster_src["month"]
        / 12
    )
)

cluster_src["month_cos"] = (
    np.cos(
        2 * np.pi
        * cluster_src["month"]
        / 12
    )
)


cluster_src["weekday_str"] = (
    cluster_src["weekday"]
    .astype(str)
)

cluster_src["label_str"] = (
    cluster_src["label"]
    .astype(str)
)


cluster_encoder = OneHotEncoder(
    sparse_output=False,
    handle_unknown="ignore"
)


cluster_categorical = (
    cluster_encoder.fit_transform(
        cluster_src[
            [
                "weekday_str",
                "label_str"
            ]
        ]
    )
)


cyclical = cluster_src[
    [
        "hour_sin",
        "hour_cos",
        "month_sin",
        "month_cos"
    ]
].to_numpy()


X_cluster = np.hstack(
    [
        cyclical,
        cluster_categorical
    ]
)


# %% 21. EVALUATE K

inertias = []
silhouettes = []
fitted_models = {}


for k in K_RANGE:

    km = KMeans(
        n_clusters=k,
        n_init=10,
        random_state=RANDOM_STATE
    )

    labels = km.fit_predict(
        X_cluster
    )

    fitted_models[k] = km

    inertias.append(
        km.inertia_
    )

    silhouettes.append(
        silhouette_score(
            X_cluster,
            labels
        )
    )


fig, axes = plt.subplots(
    1,
    2,
    figsize=(14, 5)
)


axes[0].plot(
    list(K_RANGE),
    inertias,
    marker="o"
)

axes[0].set_xlabel(
    "Number of clusters"
)

axes[0].set_ylabel(
    "Inertia"
)

axes[0].set_title(
    "Elbow plot"
)


axes[1].plot(
    list(K_RANGE),
    silhouettes,
    marker="o"
)

axes[1].set_xlabel(
    "Number of clusters"
)

axes[1].set_ylabel(
    "Silhouette score"
)

axes[1].set_title(
    "Silhouette score"
)


plt.tight_layout()
plt.show()


silhouette_table = pd.DataFrame(
    {
        "k": list(K_RANGE),
        "silhouette": silhouettes
    }
).sort_values(
    "silhouette",
    ascending=False
)


print(
    "\nSilhouette ranking:"
)

print(
    silhouette_table
    .round(4)
    .to_string(index=False)
)


TOP_KS = (
    silhouette_table
    .head(TOP_N_CLUSTER_COUNTS)["k"]
    .astype(int)
    .tolist()
)


print(
    "\nThree k-values with highest silhouette:"
)

print(
    TOP_KS
)


# Always also run k=9, in addition to the top-silhouette k-values above.
RUN_KS = sorted(set(TOP_KS) | {9})

print(
    "\nCluster solutions that will be run "
    "(top silhouette k-values + k=9):"
)

print(RUN_KS)


# %% 22. DAILY WEATHER REFERENCE FOR PERCENTILE GROUPS

unique_weather = (
    df[
        [
            "local_timestamp",
            "local_date",
            "temperature",
            "humidity",
            "snow_depth_cm",
            "precipitation"
        ]
    ]
    .drop_duplicates(
        "local_timestamp"
    )
)


daily_weather = (
    unique_weather
    .groupby("local_date")
    .agg(
        snow_depth_cm=(
            "snow_depth_cm",
            "mean"
        ),

        precipitation=(
            "precipitation",
            "sum"
        )
    )
    .reset_index()
)


# Positive snow only
snow_reference = (
    daily_weather.loc[
        daily_weather["snow_depth_cm"] > 0,
        "snow_depth_cm"
    ]
    .dropna()
)


# Positive precipitation only
precip_reference = (
    daily_weather.loc[
        daily_weather["precipitation"] > 0,
        "precipitation"
    ]
    .dropna()
)


# Hourly temperature
temperature_reference = (
    unique_weather[
        "temperature"
    ]
    .dropna()
)


# Hourly humidity
humidity_reference = (
    unique_weather[
        "humidity"
    ]
    .dropna()
)


# %% 23. PERCENTILE CLASS DEFINITIONS
#
# Unified 7-class scheme used for every variable:
# bottom 1%, top 99-90%, top 90-75%, top 75-25%,
# top 25-10%, top 10-1%, top 1%.

PERCENTILE_SPECS = [
    ("Bottom 1%", 0, 1),
    ("Top 99%-90%", 1, 10),
    ("Top 90%-75%", 10, 25),
    ("Top 75%-25%", 25, 75),
    ("Top 25%-10%", 75, 90),
    ("Top 10%-1%", 90, 99),
    ("Top 1%", 99, 100)
]

SNOW_SPECS = PERCENTILE_SPECS
PRECIP_SPECS = PERCENTILE_SPECS
TEMP_SPECS = PERCENTILE_SPECS
HUMIDITY_SPECS = PERCENTILE_SPECS


# %% 24. PRINT EXACT PERCENTILE BOUNDARIES

def percentile_table(
    reference,
    specs,
    decimals=1,
    title=""
):

    rows = []

    reference = (
        pd.Series(reference)
        .dropna()
        .astype(float)
    )


    for name, low, high in specs:

        low_value = np.percentile(
            reference,
            low
        )

        high_value = np.percentile(
            reference,
            high
        )

        rows.append(
            {
                "Class": name,
                "Percentile range":
                    f"{low}-{high}%",
                "Lower value":
                    round(
                        low_value,
                        decimals
                    ),
                "Upper value":
                    round(
                        high_value,
                        decimals
                    )
            }
        )


    table = pd.DataFrame(rows)

    print(
        f"\n{title}"
    )

    print(
        table.to_string(
            index=False
        )
    )

    return table


snow_bounds = percentile_table(
    snow_reference,
    SNOW_SPECS,
    decimals=1,
    title="SNOW DEPTH PERCENTILE BOUNDARIES"
)


precip_bounds = percentile_table(
    precip_reference,
    PRECIP_SPECS,
    decimals=1,
    title="PRECIPITATION PERCENTILE BOUNDARIES"
)


temperature_bounds = percentile_table(
    temperature_reference,
    TEMP_SPECS,
    decimals=1,
    title="TEMPERATURE PERCENTILE BOUNDARIES"
)


humidity_bounds = percentile_table(
    humidity_reference,
    HUMIDITY_SPECS,
    decimals=1,
    title="HUMIDITY PERCENTILE BOUNDARIES"
)


# %% 25. ASSIGN PERCENTILE CLASSES

def assign_percentile_class(
    values,
    reference,
    specs
):

    values = pd.Series(
        values
    )

    reference = (
        pd.Series(reference)
        .dropna()
        .astype(float)
        .sort_values()
        .to_numpy()
    )


    result = pd.Series(
        np.nan,
        index=values.index,
        dtype=object
    )


    valid = values.notna()

    valid_values = (
        values.loc[valid]
        .astype(float)
    )


    # Empirical percentile rank
    ranks = (
        np.searchsorted(
            reference,
            valid_values.to_numpy(),
            side="right"
        )
        / len(reference)
        * 100
    )


    for name, low, high in specs:

        if high == 100:

            mask = (
                ranks >= low
            ) & (
                ranks <= high
            )

        else:

            mask = (
                ranks >= low
            ) & (
                ranks < high
            )


        result.loc[
            valid_values.index[mask]
        ] = name


    return result


daily_snow_class = daily_weather.copy()

daily_snow_class[
    "snow_class"
] = assign_percentile_class(
    daily_snow_class[
        "snow_depth_cm"
    ].where(
        daily_snow_class[
            "snow_depth_cm"
        ] > 0
    ),
    snow_reference,
    SNOW_SPECS
)


daily_precip_class = daily_weather.copy()

daily_precip_class[
    "precipitation_class"
] = assign_percentile_class(
    daily_precip_class[
        "precipitation"
    ].where(
        daily_precip_class[
            "precipitation"
        ] > 0
    ),
    precip_reference,
    PRECIP_SPECS
)


# %% 26. CREATE CLUSTER PROFILE

profile = (
    df[
        [
            "label",
            "local_date",
            "hour",
            "month",
            "weekday",
            "temperature",
            "humidity",
            "is_freezing"
        ]
    ]
    .drop_duplicates(
        [
            "label",
            "local_date",
            "hour"
        ]
    )
    .copy()
)


# Merge daily classes
profile = profile.merge(
    daily_snow_class[
        [
            "local_date",
            "snow_class"
        ]
    ],
    on="local_date",
    how="left"
)


profile = profile.merge(
    daily_precip_class[
        [
            "local_date",
            "precipitation_class"
        ]
    ],
    on="local_date",
    how="left"
)


# Hourly temperature
profile[
    "temperature_class"
] = assign_percentile_class(
    profile["temperature"],
    temperature_reference,
    TEMP_SPECS
)


# Hourly humidity
profile[
    "humidity_class"
] = assign_percentile_class(
    profile["humidity"],
    humidity_reference,
    HUMIDITY_SPECS
)


# Freezing
profile[
    "freezing_class"
] = np.where(
    profile["is_freezing"] == 1,
    "Freezing",
    "Not freezing"
)


# Weekday names
weekday_names = {
    0: "Monday",
    1: "Tuesday",
    2: "Wednesday",
    3: "Thursday",
    4: "Friday",
    5: "Saturday",
    6: "Sunday"
}

profile[
    "weekday_name"
] = profile[
    "weekday"
].map(weekday_names)


# Month names
month_names = {
    1: "January",
    2: "February",
    3: "March",
    4: "April",
    5: "May",
    6: "June",
    7: "July",
    8: "August",
    9: "September",
    10: "October",
    11: "November",
    12: "December"
}

profile[
    "month_name"
] = profile[
    "month"
].map(month_names)


# %% 27. CLUSTER COMPOSITION
#
# IMPORTANT:
# normalize="columns" means:
#
# Within each percentile/category group:
#
#       cluster 0 + cluster 1 + cluster 2 + ... = 100%
#
# This directly answers:
#
# "What share of this percentile group belongs to each cluster?"
#


def cluster_composition_within_group(
    data,
    cluster_col,
    category_col,
    category_order,
    title
):

    table = pd.crosstab(
        data[cluster_col],
        data[category_col],
        normalize="columns"
    ) * 100


    table = (
        table
        .reindex(
            columns=category_order,
            fill_value=0
        )
        .round(1)
    )


    print(
        "\n"
        + "=" * 90
    )

    print(title)

    print(
        "Each column sums to approximately 100%."
    )

    print(
        "=" * 90
    )

    print(
        table.to_string()
    )

    return table


SNOW_ORDER = [
    x[0]
    for x in SNOW_SPECS
]

PRECIP_ORDER = [
    x[0]
    for x in PRECIP_SPECS
]

TEMP_ORDER = [
    x[0]
    for x in TEMP_SPECS
]

HUMIDITY_ORDER = [
    x[0]
    for x in HUMIDITY_SPECS
]

FREEZING_ORDER = [
    "Not freezing",
    "Freezing"
]

HOUR_ORDER = list(
    range(24)
)

WEEKDAY_ORDER = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

MONTH_ORDER = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June",
    "July",
    "August",
    "September",
    "October",
    "November",
    "December"
]

LINE_ORDER = LINES


# %% 28. RUN CLUSTER SOLUTIONS (TOP SILHOUETTE + k=9)

cluster_results = {}


for k in RUN_KS:

    print(
        "\n\n"
        + "#" * 100
    )

    print(
        f"CLUSTER SOLUTION k = {k}"
    )

    print(
        "#" * 100
    )


    km = fitted_models[k]

    profile_k = profile.copy()

    profile_k["cluster"] = (
        km.labels_
    )


    cluster_results[k] = (
        profile_k
    )


    silhouette_k = silhouette_score(
        X_cluster,
        km.labels_
    )


    print(
        f"\nSilhouette score: "
        f"{silhouette_k:.4f}"
    )


    # --------------------------------------------------------
    # Snow
    # --------------------------------------------------------

    cluster_composition_within_group(
        profile_k,
        "cluster",
        "snow_class",
        SNOW_ORDER,
        f"SNOW DEPTH — k={k}"
    )


    # --------------------------------------------------------
    # Precipitation
    # --------------------------------------------------------

    cluster_composition_within_group(
        profile_k,
        "cluster",
        "precipitation_class",
        PRECIP_ORDER,
        f"PRECIPITATION — k={k}"
    )


    # --------------------------------------------------------
    # Temperature
    # --------------------------------------------------------

    cluster_composition_within_group(
        profile_k,
        "cluster",
        "temperature_class",
        TEMP_ORDER,
        f"TEMPERATURE — k={k}"
    )


    # --------------------------------------------------------
    # Freezing
    # --------------------------------------------------------

    cluster_composition_within_group(
        profile_k,
        "cluster",
        "freezing_class",
        FREEZING_ORDER,
        f"FREEZING STATUS — k={k}"
    )


    # --------------------------------------------------------
    # Humidity
    # --------------------------------------------------------

    cluster_composition_within_group(
        profile_k,
        "cluster",
        "humidity_class",
        HUMIDITY_ORDER,
        f"HUMIDITY — k={k}"
    )


    # --------------------------------------------------------
    # Hour
    # --------------------------------------------------------

    cluster_composition_within_group(
        profile_k,
        "cluster",
        "hour",
        HOUR_ORDER,
        f"HOUR — k={k}"
    )


    # --------------------------------------------------------
    # Weekday
    # --------------------------------------------------------

    cluster_composition_within_group(
        profile_k,
        "cluster",
        "weekday_name",
        WEEKDAY_ORDER,
        f"WEEKDAY — k={k}"
    )


    # --------------------------------------------------------
    # Month
    # --------------------------------------------------------

    cluster_composition_within_group(
        profile_k,
        "cluster",
        "month_name",
        MONTH_ORDER,
        f"MONTH — k={k}"
    )


    # --------------------------------------------------------
    # Individual line
    # --------------------------------------------------------

    cluster_composition_within_group(
        profile_k,
        "cluster",
        "label",
        LINE_ORDER,
        f"INDIVIDUAL LINES — k={k}"
    )


# %% 29. VISUALISE CLUSTERS (CONTINUOUS MAP)
#
# A heatmap over the full hour (0-23) x month (1-12) grid, one panel
# per line, instead of a scatter cloud of individual points. Each
# cell is colored by the cluster that most often occurs at that
# hour/month for that line (there can be several calendar days
# behind one cell, so the mode is used).

def plot_cluster_map(profile_k, k):

    n_lines = len(LINES)
    ncols = 3
    nrows = int(np.ceil(n_lines / ncols))

    fig, axes = plt.subplots(
        nrows, ncols,
        figsize=(6 * ncols, 3.6 * nrows),
        squeeze=False
    )
    axes = axes.flat

    for ax, line in zip(axes, LINES):

        sub = profile_k[profile_k["label"] == line]

        if sub.empty:
            ax.set_axis_off()
            continue

        grid = (
            sub.groupby(["month", "hour"])["cluster"]
            .agg(lambda x: x.mode().iloc[0])
            .unstack("hour")
            .reindex(index=range(1, 13), columns=range(24))
        )

        sns.heatmap(
            grid,
            cmap="tab10",
            vmin=0,
            vmax=max(profile_k["cluster"].max(), 1),
            cbar_kws={"label": "Cluster"},
            linewidths=0.3,
            linecolor="white",
            ax=ax
        )

        ax.set_title(line)
        ax.set_xlabel("Hour")
        ax.set_ylabel("Month")

    for ax in list(axes)[n_lines:]:
        ax.set_axis_off()

    fig.suptitle(f"Dominant cluster by hour and month - k={k}", y=1.02, fontsize=15)
    plt.tight_layout()
    plt.show()


for k, profile_k in cluster_results.items():
    plot_cluster_map(profile_k, k)


# %% 30. DELAY BY CLUSTER

for k, profile_k in cluster_results.items():

    assignment = (
        profile_k[
            [
                "label",
                "local_date",
                "hour",
                "cluster"
            ]
        ]
        .drop_duplicates()
    )


    delay_cluster = df.merge(
        assignment,
        on=[
            "label",
            "local_date",
            "hour"
        ],
        how="left"
    )


    summary = (
        delay_cluster
        .groupby("cluster")[TARGET]
        .agg(
            mean="mean",
            median="median",
            count="count"
        )
        .round(2)
    )


    print(
        "\n"
        + "=" * 80
    )

    print(
        f"MEDIAN DELAY BY CLUSTER — k={k}"
    )

    print(
        "=" * 80
    )

    print(
        summary.to_string()
    )


    plt.figure(
        figsize=(9, 5)
    )

    sns.boxplot(
        data=delay_cluster.dropna(
            subset=["cluster"]
        ),
        x="cluster",
        y=TARGET
    )

    plt.title(
        f"Median delay by cluster - k={k}"
    )

    plt.xlabel(
        "Cluster"
    )

    plt.ylabel(
        "Median delay"
    )

    plt.tight_layout()
    plt.show()
```

```python
# ============================================================
# MARKENS TILLSTÅND / RÅDANDE VÄDER — LARGE BOXPLOTS ONLY
#
# A small, SELF-CONTAINED companion script to the main analysis.
# It only makes the two categorical boxplots (soil condition,
# prevailing weather) against median_delay, but bigger and sorted
# so it's actually possible to see which categories contribute
# the most delay.
#
# This script does its own loading and merging of the three raw
# files (modelling data + the two SMHI files) - it does NOT
# depend on the main script having been run first, and does not
# read "modelling_data_2024_with_SMHI.csv".
#
# Changes vs. the boxplots in the main script:
#   - Each category gets its own full-size figure instead of
#     being squeezed into a 2-panel row - much more vertical
#     room on the median_delay axis.
#   - Horizontal orientation (category on the y-axis) so long
#     Swedish weather descriptions are readable without rotated
#     text, and figure height grows with the number of categories.
#   - Categories are ordered by descending median delay, so the
#     biggest contributors to delay are immediately visible at
#     the top of each plot.
# ============================================================


# %% 1. IMPORTS AND SETTINGS

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

MODELLING_FILE = "modelling_data_2024.csv"

SOIL_FILE = (
    "(markens tillstånd_dag Vallentuna)"
    " smhi-opendata_40_98310_202401_202412.csv"
)

WEATHER_FILE = (
    "(rådande väder Bromma)"
    " smhi-opendata_13_97200_202401_202412.csv"
)

TARGET = "median_delay"

LINES = [
    "BUS 1",
    "BUS 4",
    "METRO 10",
    "METRO 19",
    "TRAIN 43",
    "TRAM 30"
]

# Same ~1800 outlier removal as the main script - only affects
# these plots, not the underlying data.
BOXPLOT_OUTLIER_TARGET = 1800
BOXPLOT_OUTLIER_TOLERANCE = 100


# %% 2. HELPERS

def read_smhi(path):
    """
    Read an SMHI CSV and automatically locate the row
    beginning with 'Datum'.
    """
    with open(path, encoding="utf-8-sig") as f:
        header_row = next(
            i for i, line in enumerate(f)
            if line.startswith("Datum")
        )

    out = pd.read_csv(
        path,
        skiprows=header_row,
        sep=None,
        engine="python",
        encoding="utf-8-sig"
    )
    out.columns = out.columns.str.strip()
    return out


def utc_to_local(date_col, time_col):
    """
    Convert SMHI UTC timestamp to Europe/Stockholm.
    """
    utc = pd.to_datetime(
        date_col.astype(str) + " " + time_col.astype(str),
        errors="coerce"
    )
    return (
        utc.dt.tz_localize("UTC")
        .dt.tz_convert("Europe/Stockholm")
        .dt.tz_localize(None)
    )


def numeric_code(value):
    try:
        return int(float(value))
    except (ValueError, TypeError):
        return None


def decode_category(value, mapping, fallback=True):
    code = numeric_code(value)
    if code is None:
        return str(value)
    if code in mapping:
        return mapping[code]
    if fallback:
        return f"Unknown code {code}"
    return str(value)


# %% 3. RÅDANDE VÄDER CODE MAP
#
# Same mapping as the main script.

RADANDE_VADER_MAP = {

    0: "Molnens utveckling har icke kunnat observeras eller icke observerats",
    1: "Moln har upplösts helt eller avtagit i utsträckning, i mäktighet eller i täthet",
    2: "Molnhimlen i stort sett oförändrad",
    3: "Moln har bildats eller tilltagit i utsträckning, i mäktighet eller i täthet",
    4: "Sikten nedsatt av brandrök eller fabriksrök",
    5: "Torrdis (solrök)",
    6: "Stoft svävar i luften, men uppvirvlas ej av vinden vid observationsterminen",
    7: "Stoft eller sand uppvirvlas av vinden men ingen utpräglad sandvirvel och ingen sandstorm inom synhåll",
    8: "Utpräglad stoft- eller sandvirvel vid observationsterminen eller under senaste timmen men ingen sandstorm",
    9: "Sandstorm under senaste timmen eller inom synhåll vid observationsterminen",
    10: "Fuktdis med sikt 1-10 km",
    11: "Låg dimma i bankar på stationen",
    12: "Mer eller mindre sammanhängande låg dimma på stationen",
    13: "Kornblixt",
    14: "Nederbörd inom synhåll, som ej når marken eller havsytan",
    15: "Nederbörd som når marken eller havsytan på avstånd större än 5 km",
    16: "Nederbörd som når marken eller havsytan på avstånd mindre än 5 km",
    17: "Åska vid observationsterminen men ingen nederbörd på stationen",
    18: "Utpräglade starka vindbyar",
    19: "Skydrag eller tromb",
    20: "Duggregn eller kornsnö under senaste timmen men ej vid observationsterminen",
    21: "Regn under senaste timmen men ej vid observationsterminen",
    22: "Snöfall under senaste timmen men ej vid observationsterminen",
    23: "Snöblandat regn eller iskorn under senaste timmen men ej vid observationsterminen",
    24: "Underkylt regn eller duggregn under senaste timmen men ej vid observationsterminen",
    25: "Regnskurar under senaste timmen men ej vid observationsterminen",
    26: "Byar av snö eller snöblandat regn under senaste timmen men ej vid observationsterminen",
    27: "Byar av hagel med eller utan regn under senaste timmen men ej vid observationsterminen",
    28: "Dimma under senaste timmen men ej vid observationsterminen",
    29: "Åska med eller utan nederbörd under senaste timmen men ej vid observationsterminen",
    30: "Lätt eller måttlig sandstorm har avtagit i styrka",
    31: "Lätt eller måttlig sandstorm utan märkbar förändring",
    32: "Lätt eller måttlig sandstorm har börjat eller tilltagit i styrka",
    33: "Kraftig sandstorm har avtagit i styrka",
    34: "Kraftig sandstorm utan märkbar förändring",
    35: "Kraftig sandstorm har börjat eller tilltagit i styrka",
    36: "Lågt och lätt eller måttligt snödrev",
    37: "Lågt men tätt snödrev",
    38: "Högt men lätt eller måttligt snödrev",
    39: "Högt och tätt snödrev",
    40: "Dimma inom synhåll vid observationsterminen, nående över ögonhöjd",
    41: "Dimma i bankar på stationen",
    42: "Dimma, med skymt av himlen, har blivit lättare",
    43: "Dimma, utan skymt av himlen, har blivit lättare",
    44: "Dimma, med skymt av himlen, oförändrad",
    45: "Dimma, utan skymt av himlen, oförändrad",
    46: "Dimma, med skymt av himlen, har börjat eller tätnat",
    47: "Dimma, utan skymt av himlen, har börjat eller tätnat",
    48: "Underkyld dimma, med skymt av himlen",
    49: "Underkyld dimma, utan skymt av himlen",

    50: "Lätt duggregn med avbrott",
    51: "Lätt duggregn, ihållande",
    52: "Måttligt duggregn med avbrott",
    53: "Måttligt duggregn, ihållande",
    54: "Tätt duggregn med avbrott",
    55: "Tätt duggregn, ihållande",
    56: "Lätt underkylt duggregn",
    57: "Måttligt eller tätt underkylt duggregn",
    58: "Lätt duggregn tillsammans med regn",
    59: "Måttligt eller tätt duggregn tillsammans med regn",

    60: "Lätt regn med avbrott",
    61: "Lätt regn, ihållande",
    62: "Måttligt regn med avbrott",
    63: "Måttligt regn, ihållande",
    64: "Starkt regn med avbrott",
    65: "Starkt regn ihållande",
    66: "Lätt underkylt regn",
    67: "Måttligt eller starkt underkylt regn",
    68: "Lätt regn eller duggregn tillsammans med snö",
    69: "Måttligt eller starkt regn eller duggregn tillsammans med snö",

    70: "Lätt snöfall med avbrott",
    71: "Lätt snöfall, ihållande",
    72: "Måttligt snöfall med avbrott",
    73: "Måttligt snöfall, ihållande",
    74: "Tätt snöfall med avbrott",
    75: "Tätt snöfall, ihållande",
    76: "Isnålar",
    77: "Kornsnö",
    78: "Enstaka snöstjärnor",
    79: "Iskorn",

    80: "Lätta regnskurar",
    81: "Måttliga eller kraftiga regnskurar",
    82: "Mycket kraftiga regnskurar (skyfall)",
    83: "Lätt snöblandat regn i byar",
    84: "Måttligt eller kraftigt snöblandat regn i byar",
    85: "Lätta snöbyar",
    86: "Måttliga eller kraftiga snöbyar",
    87: "Lätta byar av småhagel eller snöhagel",
    88: "Måttliga eller kraftiga byar av småhagel eller snöhagel",
    89: "Lätta byar av ishagel utan åska",
    90: "Måttliga eller kraftiga byar av ishagel utan åska",

    91: "Lätt regn vid observationsterminen, åskväder under senaste timmen",
    92: "Måttligt eller starkt regn vid observationsterminen, åskväder under senaste timmen",
    93: "Lätt snöfall, snöblandat regn eller hagel vid observationsterminen, åskväder senaste timmen",
    94: "Måttligt eller starkt snöfall, snöblandat regn eller hagel, åskväder senaste timmen",

    95: "Svagt eller måttligt åskväder utan hagel men med regn eller snö",
    96: "Svagt eller måttligt åskväder med hagel",
    97: "Kraftigt åskväder utan hagel men med regn eller snö",
    98: "Kraftigt åskväder med sandstorm",
    99: "Kraftigt åskväder med hagel",

    100: "Inget signifikant väder observerat",
    101: "Moln har upplösts helt eller avtagit under senaste timmen",
    102: "Molnhimlen i stort sett oförändrad under senaste timmen",
    103: "Moln har bildats eller tilltagit under senaste timmen",
    104: "Dis eller rök eller stoft i luften, sikt >= 1 km",
    105: "Dis eller rök eller stoft i luften, sikt < 1 km",

    110: "Fuktdis med sikt 1-10 km",
    111: "Isnålar",
    112: "Blixt på avstånd",
    118: "Utpräglade starka vindbyar",

    120: "Dimma",
    121: "Nederbörd",
    122: "Duggregn eller kornsnö",
    123: "Regn",
    124: "Snöfall",
    125: "Underkylt duggregn eller regn",
    126: "Åskväder med eller utan nederbörd",
    127: "Snödrev eller sandstorm",
    128: "Snödrev eller sandstorm, sikt >= 1 km",
    129: "Snödrev eller sandstorm, sikt < 1 km",
    130: "Dimma",
    131: "Dimma i bankar på stationen",
    132: "Dimma har blivit lättare under senaste timmen",
    133: "Dimma oförändrad under senaste timmen",
    134: "Dimma har börjat eller tätnat under senaste timmen",
    135: "Underkyld dimma",

    140: "Nederbörd",
    141: "Lätt eller måttlig nederbörd",
    142: "Kraftig nederbörd",
    143: "Flytande nederbörd, lätt eller måttlig",
    144: "Flytande nederbörd, kraftig",
    145: "Fast nederbörd, lätt eller måttlig",
    146: "Fast nederbörd, kraftig",
    147: "Lätt eller måttlig underkyld nederbörd",
    148: "Kraftig underkyld nederbörd",

    150: "Duggregn",
    151: "Lätt duggregn",
    152: "Måttligt duggregn",
    153: "Tätt duggregn",
    154: "Lätt underkylt duggregn",
    155: "Måttligt underkylt duggregn",
    156: "Tätt underkylt duggregn",
    157: "Lätt duggregn tillsammans med regn",
    158: "Måttligt eller tätt duggregn tillsammans med regn",

    160: "Regn",
    161: "Lätt regn",
    162: "Måttligt regn",
    163: "Starkt regn",
    164: "Lätt underkylt regn",
    165: "Måttligt underkylt regn",
    166: "Starkt underkylt regn",
    167: "Lätt regn eller duggregn tillsammans med snö",
    168: "Måttligt eller starkt regn eller duggregn tillsammans med snö",

    170: "Snöfall",
    171: "Lätt snöfall",
    172: "Måttligt snöfall",
    173: "Tätt snöfall",
    174: "Lätt småhagel",
    175: "Måttligt småhagel",
    176: "Kraftigt småhagel",
    177: "Kornsnö",
    178: "Isnålar",

    180: "Regnskurar",
    181: "Lätta regnskurar",
    182: "Måttliga regnskurar",
    183: "Kraftiga regnskurar",
    184: "Mycket kraftiga regnskurar (skyfall)",
    185: "Lätta snöbyar",
    186: "Måttliga snöbyar",
    187: "Kraftiga snöbyar",
    189: "Hagel",

    190: "Åskväder",
    191: "Svagt eller måttligt åskväder utan nederbörd",
    192: "Svagt eller måttligt åskväder med regnskurar eller snöbyar",
    193: "Svagt eller måttligt åskväder med hagel",
    194: "Kraftigt åskväder utan nederbörd",
    195: "Kraftigt åskväder med regnskurar eller snöbyar",
    196: "Kraftigt åskväder med hagel",
    199: "Tromb eller tornado",

    204: "Vulkanaska som spridits högt upp i luften",
    206: "Tjockt stoftdis, sikt < 1 km",
    207: "Vattenstänk vid station pga blåst",
    208: "Drivande stoft eller sand",
    209: "Kraftig stoft- eller sandstorm på avstånd",
    210: "Snödis",
    211: "Snöstorm eller kraftigt snödrev som ger extremt dålig sikt",
    213: "Blixt mellan moln och marken",
    217: "Åska utan regnskur",
    219: "Tromb eller tornado vid stationen eller inom synhåll under senaste timmen",

    220: "Avlagring av vulkanaska",
    221: "Avlagring av stoft eller sand",
    222: "Dagg",
    223: "Utfällning av blöt snö",
    224: "Lätt eller måttlig dimfrost",
    225: "Kraftig dimfrost",
    226: "Rimfrost",
    227: "Kraftig isbeläggning pga underkyld nederbörd",
    228: "Isskorpa",

    230: "Stoft- eller sandstorm med temperatur under fryspunkten",
    239: "Kraftigt snödrev och/eller snöfall",

    241: "Dimma till havs",
    242: "Dimma i dalgång",
    243: "Sjörök i Arktis eller vid Antarktis",
    244: "Advektionsdimma över vatten",
    245: "Advektionsdimma över land",
    246: "Dimma över is eller snö",
    247: "Tät dimma, sikt 60-90 m",
    248: "Tät dimma, sikt 30-60 m",
    249: "Tät dimma, sikt < 30 m",

    250: "Duggregn, intensitet < 0,10 mm/timme",
    251: "Duggregn, intensitet 0,10-0,19 mm/timme",
    252: "Duggregn, intensitet 0,20-0,39 mm/timme",
    253: "Duggregn, intensitet 0,40-0,79 mm/timme",
    254: "Duggregn, intensitet 0,80-1,59 mm/timme",
    255: "Duggregn, intensitet 1,60-3,19 mm/timme",
    256: "Duggregn, intensitet 3,20-6,39 mm/timme",
    257: "Duggregn, intensitet > 6,40 mm/timme",

    259: "Duggregn och snöfall",

    260: "Regn, intensitet < 1,0 mm/timme",
    261: "Regn, intensitet 1,0-1,9 mm/timme",
    262: "Regn, intensitet 2,0-3,9 mm/timme",
    263: "Regn, intensitet 4,0-7,9 mm/timme",
    264: "Regn, intensitet 8,0-15,9 mm/timme",
    265: "Regn, intensitet 16,0-31,9 mm/timme",
    266: "Regn, intensitet 32,0-63,9 mm/timme",
    267: "Regn, intensitet > 64,0 mm/timme",

    270: "Snö, intensitet < 1,0 cm/timme",
    271: "Snö, intensitet 1,0-1,9 cm/timme",
    272: "Snö, intensitet 2,0-3,9 cm/timme",
    273: "Snö, intensitet 4,0-7,9 cm/timme",
    274: "Snö, intensitet 8,0-15,9 cm/timme",
    275: "Snö, intensitet 16,0-31,9 cm/timme",
    276: "Snö, intensitet 32,0-63,9 cm/timme",
    277: "Snö, intensitet > 64,0 cm/timme",

    278: "Snöfall eller isnålar från klar himmel",
    279: "Frysande blötsnö",
    280: "Regn",
    281: "Underkylt regn",
    282: "Snöblandat regn",
    283: "Snöfall",
    284: "Småhagel eller snöhagel",
    285: "Småhagel eller snöhagel tillsammans med regn",
    286: "Småhagel eller snöhagel tillsammans med snöblandat regn",
    287: "Småhagel eller snöhagel tillsammans med snö",
    288: "Hagel",
    289: "Hagel tillsammans med regn",
    290: "Hagel tillsammans med snöblandat regn",
    291: "Hagel tillsammans med snö",
    292: "Skurar eller åska till havs",
    293: "Skurar eller åska över berg",

    508: "Inga signifikanta fenomen att rapportera",
    509: "Ingen observation, data ej tillgängligt",
    510: "Rådande och gammalt väder saknas men förväntades",
    511: "Saknat värde"
}


# %% 4. MARKENS TILLSTÅND MAP
#
# Same as the main script - fill in codes here if you have a
# mapping table for soil condition.

MARKENS_TILLSTAND_MAP = {}


# %% 5. LOAD + MERGE DATA
#
# Only the two columns these plots actually need - soil condition
# (daily) and prevailing weather (hourly) - are merged in, to
# keep this self-contained version lean.

df = pd.read_csv(MODELLING_FILE)
df.columns = df.columns.str.strip()

df["date_dt"] = pd.to_datetime(df["date"].astype(str), format="%Y%m%d")
df["local_timestamp"] = df["date_dt"] + pd.to_timedelta(df["hour"], unit="h")
df["local_date"] = df["date_dt"].dt.date


# Soil condition (daily)
soil = read_smhi(SOIL_FILE).rename(
    columns={"Datum": "local_date", "Markens tillstånd": "markens_tillstand"}
)
soil["local_date"] = pd.to_datetime(soil["local_date"]).dt.date
soil["markens_tillstand"] = soil["markens_tillstand"].astype(str).str.strip()
soil = soil[["local_date", "markens_tillstand"]].drop_duplicates("local_date")


# Prevailing weather (hourly)
weather = read_smhi(WEATHER_FILE)
weather["local_timestamp"] = utc_to_local(weather["Datum"], weather["Tid (UTC)"])
weather["radande_vader"] = weather["Rådande väder"].astype(str).str.strip()
weather = (
    weather.groupby("local_timestamp", as_index=False)["radande_vader"]
    .agg(lambda x: x.mode().iloc[0])
)


df = (
    df
    .merge(weather, on="local_timestamp", how="left", validate="many_to_one")
    .merge(soil, on="local_date", how="left", validate="many_to_one")
)

print("Rows:", len(df))
print("\nMissing values:")
print(df[["radande_vader", "markens_tillstand"]].isna().sum())


# %% 6. REMOVE THE ~1800 OUTLIER (BOXPLOTS ONLY)

valid_delay = df[TARGET].dropna()
closest_idx = (valid_delay - BOXPLOT_OUTLIER_TARGET).abs().idxmin()
closest_value = df.loc[closest_idx, TARGET]

if abs(closest_value - BOXPLOT_OUTLIER_TARGET) <= BOXPLOT_OUTLIER_TOLERANCE:
    BOX_OUTLIER_LABEL = {
        "label": df.loc[closest_idx, "label"],
        "timestamp": df.loc[closest_idx, "local_timestamp"]
    }
    print("\nRemoved from boxplots:", BOX_OUTLIER_LABEL, "value:", closest_value)
else:
    BOX_OUTLIER_LABEL = None
    print("\nNo ~1800 outlier found within tolerance.")


# %% 7. DECODE CATEGORICAL VALUES

df["weather_label"] = df["radande_vader"].map(
    lambda x: decode_category(x, RADANDE_VADER_MAP)
)
df["soil_label"] = df["markens_tillstand"].map(
    lambda x: decode_category(x, MARKENS_TILLSTAND_MAP)
)


# %% 8. BIG, SORTED BOXPLOT FUNCTION

def big_categorical_boxplot(line_df, category_col, name, title):
    """
    One large horizontal boxplot: category on the y-axis (so long
    labels stay readable without rotation), median_delay on the
    x-axis (so there's room to actually see the spread), and
    categories sorted by descending median delay so the biggest
    contributors appear at the top.
    """

    plot_df = line_df.copy()

    if BOX_OUTLIER_LABEL is not None:
        plot_df = plot_df[
            ~(
                (plot_df["label"] == BOX_OUTLIER_LABEL["label"])
                & (plot_df["local_timestamp"] == BOX_OUTLIER_LABEL["timestamp"])
            )
        ]

    data = plot_df[[category_col, TARGET]].dropna()

    if data.empty:
        print(f"No data for {title} - skipped.")
        return

    # Sort categories by descending median delay - the categories
    # that contribute the most delay show up first.
    order = (
        data.groupby(category_col)[TARGET]
        .median()
        .sort_values(ascending=False)
        .index
        .tolist()
    )

    n_categories = len(order)
    fig_height = max(5, 0.45 * n_categories)

    plt.figure(figsize=(14, fig_height))

    sns.boxplot(
        data=data,
        x=TARGET,
        y=category_col,
        order=order,
        orient="h"
    )

    plt.title(f"{title} (sorted by median delay, highest first)", fontsize=14)
    plt.xlabel("Median delay")
    plt.ylabel("")
    plt.grid(axis="x", alpha=0.4)
    plt.tight_layout()
    plt.show()

    # Also print the sorted medians as a table, for exact values
    # that can be hard to read off a crowded plot.
    summary = (
        data.groupby(category_col)[TARGET]
        .agg(median="median", mean="mean", count="count")
        .reindex(order)
        .round(2)
    )
    print(f"\n{title} - sorted by median delay:")
    print(summary.to_string())


# %% 9. RUN: ALL LINES

big_categorical_boxplot(df, "soil_label", "ALL LINES", "Markens tillstånd vs median_delay")
big_categorical_boxplot(df, "weather_label", "ALL LINES", "Rådande väder vs median_delay")


# %% 10. RUN: EACH LINE

for line in LINES:
    line_df = df[df["label"] == line]
    big_categorical_boxplot(line_df, "soil_label", line, f"Markens tillstånd vs median_delay - {line}")
    big_categorical_boxplot(line_df, "weather_label", line, f"Rådande väder vs median_delay - {line}")
```
